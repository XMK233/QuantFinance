# 单线程模式每日交易系统

专门为解决 baostock 多进程访问限制问题设计的运行环境。

## 问题背景

baostock 对多进程并发访问有限制，当使用多进程模式时可能会遇到：
- 网络接收错误
- 连接失败
- 下载超时

单线程模式可以避免这些问题，提供最稳定的数据下载体验。

## 配置参数

修改以下参数以适应您的需求：

```python
# 并发模式配置
MODE = 'single'  # 固定为单线程模式
WORKERS = 1      # 固定为1个工作线程

# 策略配置
STRATEGIES = ['weekly_mean_down', 'cross_ma50']  # 使用的策略
PRICE_CAP = 150.0  # 价格上限
MAX_HOLDINGS = 3   # 最大持仓数

# 数据更新配置
RECOMMEND_ONLY = False  # 是否只生成推荐
SKIP_UPDATE = False     # 是否跳过数据更新
UPDATE_ONLY = False     # 是否只更新数据

# 其他配置
IGNORE_HOLDINGS = False  # 是否忽略持仓限制
CLEAR_POSITIONS = False  # 是否清空持仓记录
```

## 使用方法

### 1. 完整运行（更新数据 + 生成推荐）

```python
print("🚀 开始完整运行（单线程模式）...")
print("=" * 60)

!python daily_trading_system.py \
  --mode {MODE} \
  --workers {WORKERS} \
  --strategy {' '.join(STRATEGIES)} \
  --price-cap {PRICE_CAP} \
  --max-holdings {MAX_HOLDINGS}
```

### 2. 只更新数据

```python
print("📊 只更新数据（单线程模式）...")
print("=" * 60)

!python daily_trading_system.py \
  --mode {MODE} \
  --workers {WORKERS} \
  --update-only
```

## 故障排除

### 常见问题及解决方案

| 问题 | 可能原因 | 解决方案 |
|------|----------|----------|
| 网络接收错误 | baostock 服务器限制 | 使用单线程模式，降低请求频率 |
| 连接失败 | 网络问题或服务器维护 | 检查网络连接，稍后再试 |
| 下载超时 | 单个股票下载时间过长 | 增加超时时间，或跳过该股票 |
| 数据不完整 | 数据库连接问题 | 检查数据库文件路径和权限 |

## 总结

单线程模式提供了最稳定的 baostock 数据下载体验，特别适合：

1. **网络环境不稳定**时
