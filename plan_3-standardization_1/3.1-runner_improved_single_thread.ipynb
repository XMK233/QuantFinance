# 改进版每日交易系统Runner - 支持单线程参数

## 背景
由于baostock对多进程/多线程访问有限制，本notebook提供了单线程模式的支持，避免因并发访问导致的网络错误。

## 并发模式说明
- **single**: 单线程模式 - 避免baostock限制，稳定但较慢
- **thread**: 多线程模式 - 中等并发，适用于一般场景
- **process**: 多进程模式 - 高并发，默认模式（但可能触发baostock限制）

## 配置参数
可以根据需要调整以下参数：

```python
# 配置参数
CONCURRENCY_MODE = "single"  # 单线程模式，避免baostock限制
WORKERS = 1                   # 工作线程数（单线程模式建议设为1）
STRATEGIES = ["weekly_mean_down", "cross_ma50"]  # 使用的策略
PRICE_CAP = 150.0            # 价格上限
MAX_HOLDINGS = 3             # 最大持仓数量
INITIAL_CAPITAL = 50000      # 初始资金
RECOMMEND_ONLY = True        # 仅推荐模式（不实际交易）
```

## 1. 数据更新（单线程模式）

使用单线程模式更新数据，避免baostock并发限制：

```python
# 更新每日数据（单线程模式）
!python daily_update.py --mode {CONCURRENCY_MODE} --workers {WORKERS}
```

## 2. 交易推荐（单线程模式）

使用单线程模式进行交易推荐：

```python
# 生成交易推荐（单线程模式）
!python daily_trading_system.py \
  --mode {CONCURRENCY_MODE} \
  --workers {WORKERS} \
  --strategy {' '.join(STRATEGIES)} \
  --price-cap {PRICE_CAP} \
  --max-holdings {MAX_HOLDINGS} \
  --initial-capital {INITIAL_CAPITAL} \
  --recommend-only
```

## 使用指南

### 1. 首次使用
